```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("climate_change_dataset.csv")
df_2023 = df[df['Year'] == 2023]

countries = ['South Africa','India','France','Indonesia','USA','Argentina']
df_6=df_2023[df_2023['Country'].isin(countries)]


plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df_6,
    x='CO2 Emissions (Tons/Capita)',
    y='Avg Temperature (°C)',
    hue='Country',
    alpha=0.8
)

plt.title('CO2 Emissions vs Average Temperature (2023) [SCATTER PLOT]')
plt.xlabel('CO2 Emissions (Tons/Capita)')
plt.ylabel('Average Temperature (°C)')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("climate_change_dataset.csv")
df_2023 = df[df['Year'] == 2023]

countries = df_2023['Country'].unique()[:6]
df_6=df_2023[df_2023['Country'].isin(countries)]

co2_by_country = (df_6.groupby('Country')['CO2 Emissions (Tons/Capita)'].sum() .nlargest(6))

plt.figure(figsize=(7, 7))

plt.pie(
    co2_by_country,
    labels=co2_by_country.index,
    autopct='%1.1f%%',
    startangle=140
)

plt.title('CO2 Emissions (2023) [PIE CHART]')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("climate_change_dataset.csv")
df_2023 = df[df['Year'] == 2023]

countries = df_2023['Country'].unique()[:6]
df_6=df_2023[df_2023['Country'].isin(countries)]

#scatter plot
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df_6,
    x='CO2 Emissions (Tons/Capita)',
    y='Avg Temperature (°C)',
    hue='Country',
    alpha=0.8
)

plt.title('CO2 Emissions vs Average Temperature (2023) [SCATTER PLOT]')
plt.xlabel('CO2 Emissions (Tons/Capita)')
plt.ylabel('Average Temperature (°C)')

plt.tight_layout()
plt.show()

#line plot
plt.figure(figsize=(8, 5))
sns.lineplot(
    data=df,
    x='Year',
    y='Avg Temperature (°C)',
    marker="o"
)

plt.title('Average Temperature Trend (2000 - 2023) [LINE CHART]')
plt.xlabel('Year')
plt.ylabel('Average Temperature (°C)')

plt.show()

#barplot
plt.figure(figsize=(8, 5))
sns.barplot( data=df_6, x='Country',y='Avg Temperature (°C)',errorbar=None )

plt.title('Average Temperature by Country (2023) [BAR PLOT]')
plt.xlabel('Country')
plt.ylabel('Average Temperature (°C)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

#boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(data=df_6, x='Country', y='Avg Temperature (°C)')

plt.title('Distribution of Average Temperature by Country (2023) [BOX PLOT]')
plt.xlabel('Country')
plt.ylabel('Average Temperature (°C)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

#piechart
co2_by_country = (df_6.groupby('Country')['CO2 Emissions (Tons/Capita)'].sum() .nlargest(6))

plt.figure(figsize=(7, 7))

plt.pie(
    co2_by_country,
    labels=co2_by_country.index,
    autopct='%1.1f%%',
    startangle=140
)

plt.title('CO2 Emissions (2023) [PIE CHART]')
plt.show()

#histogram
plt.figure(figsize=(8, 5))
sns.histplot(data=df_2023, x='Population', bins=10)

plt.title('Global Distribution of Population (2023) [HISTOGRAM]')
plt.xlabel('Population (Years)')
plt.ylabel('Frequency (Number of Countries)')
plt.show()
```
