```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("student_dropout_behavior_dataset.csv")

x = data["quiz1_marks"]
y = data["final_marks"]

covariance = x.cov(y)
correlation = x.corr(y)

print("----- Correlation and Covariance Analysis -----")
print("Independent Variable (X): quiz1_marks")
print("Dependent Variable (Y): final_marks")
print("Number of observations:", len(data))

print("\nCovariance =", covariance)
print("Correlation Coefficient (r) =", correlation)

if abs(correlation) >= 0.7:
    strength = "strong"
elif abs(correlation) >= 0.4:
    strength = "moderate"
elif abs(correlation) >= 0.2:
    strength = "weak"
else:
    strength = "very weak"

if correlation > 0:
    direction = "positive"
elif correlation < 0:
    direction = "negative"
else:
    direction = "no"

print("\nAnalysis:")
print("The correlation is", direction, "and", strength + ".")
print("This indicates how Quiz 1 Marks are linearly related to Final Marks.")

plt.scatter(x, y)
plt.xlabel("Quiz 1 Marks")
plt.ylabel("Final Marks")
plt.title("Quiz 1 Marks vs Final Marks")
plt.grid(True)
plt.show()
```
