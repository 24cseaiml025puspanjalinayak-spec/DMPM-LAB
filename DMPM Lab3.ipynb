```python
import numpy as np
data = np.array([10, 20, 30, 40, 50])

mean = np.mean(data)
std = np.std(data)

z_score = (data - mean) / std

min_val = np.min(data)
max_val = np.max(data)

min_max = (data - min_val) / (max_val - min_val)

j = len(str(max(abs(data))))
decimal_scaling = data / (10 ** j)

print("Original Data:       ", data)
print("Z-Score:             ", np.round(z_score, 3))
print("Min-Max:             ", np.round(min_max, 2))
print("Decimal Scaling:     ", np.round(decimal_scaling, 2))
```

```python
import pandas as pd
import numpy as np

data = pd.read_csv("score_updated.csv")
print("Original Dataset:")
print(data.head())
numeric_columns = data.select_dtypes(include=np.number).columns

normalized_data = data.copy()
for column in numeric_columns:
    max_value = data[column].abs().max()
    if max_value == 0:
        j = 0
    else:
        j = len(str(int(max_value)))
    normalized_data[column] = data[column] / (10 ** j)

print("\nDecimal Scaling Normalized Dataset:")
print(normalized_data.head())
```

```python
import pandas as pd
import numpy as np
data = pd.read_csv("score_updated.csv")

numeric_columns = data.select_dtypes(include='number').columns
normalized_data = data.copy()
for column in numeric_columns:
    min_value = data[column].min()
    max_value = data[column].max()

    normalized_data[column] = (
        (data[column] - min_value) / (max_value - min_value)
    )
print("Min-Max Normalized Dataset:")
display(normalized_data)
```

```python
import pandas as pd
import numpy as np

# Read dataset
data = pd.read_csv("score_updated.csv")

# Select numeric columns
numeric_columns = data.select_dtypes(include=np.number).columns

# Calculate mean, variance and standard deviation
mean = data[numeric_columns].mean()
variance = data[numeric_columns].var()
std = data[numeric_columns].std()

print("MEAN:",mean)

print("\nVARIANCE:")
print(variance)

print("\nSTANDARD DEVIATION:")
print(std)

# Z-score normalization
zscore_data = data.copy()

for column in numeric_columns:
    zscore_data[column] = (data[column] - mean[column]) / std[column]

# Print only first 10 rows
print("\nZ-SCORE NORMALIZED DATA:")
print(zscore_data.head(10))
```
